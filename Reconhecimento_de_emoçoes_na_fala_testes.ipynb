```python
# Importar as bibliotecas
import yt_dlp
from pydub import AudioSegment
import whisper
import os

# Função para baixar o vídeo do YouTube
def download_audio_from_youtube(video_url, output_filename='audio.mp3'):
    try:
        ydl_opts = {
            'format': 'bestaudio/best',
            'outtmpl': output_filename,
            'postprocessors': [{
                'key': 'FFmpegExtractAudio',
                'preferredcodec': 'mp3',
                'preferredquality': '192',
            }],
        }
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([video_url])
        print(f"Áudio baixado com sucesso: {output_filename}")
    except Exception as e:
        print(f"Erro ao baixar o vídeo: {e}")

# Função para converter o áudio para WAV
def convert_audio_to_wav(input_filename, output_filename='audio.wav'):
    try:
        audio = AudioSegment.from_file(input_filename)
        audio.export(output_filename, format='wav')
        print(f"Áudio convertido com sucesso: {output_filename}")
    except Exception as e:
        print(f"Erro ao converter o áudio: {e}")

# Função para transcrever o áudio para texto
def transcribe_audio_to_text(audio_filename, output_text_filename='transcription.txt'):
    try:
        model = whisper.load_model("base")
        result = model.transcribe(audio_filename)
        with open(output_text_filename, 'w') as f:
            f.write(result['text'])
        print(f"Transcrição concluída e salva em '{output_text_filename}'")
    except Exception as e:
        print(f"Erro ao transcrever o áudio: {e}")

# URL do vídeo do YouTube
video_url = 'https://www.youtube.com/watch?v=mWD8wWwZpi8'

# Passo 1: Baixar o vídeo do YouTube
download_audio_from_youtube(video_url, 'audio.mp3')

# Verificar se o arquivo baixado existe e ajustar o nome do arquivo, se necessário
if not os.path.exists('audio.mp3') and os.path.exists('audio.mp3.mp3'):
    os.rename('audio.mp3.mp3', 'audio.mp3')

# Passo 2: Converter o áudio para WAV
convert_audio_to_wav('audio.mp3', 'audio.wav')

# Passo 3: Transcrever o áudio para texto
transcribe_audio_to_text('audio.wav')
```

```python
import os
import pandas as pd

# 1. Defina o caminho para sua pasta EXTRAÍDA
caminho_da_pasta_principal = '/content/drive/MyDrive/Colab Notebooks/Atividades Pos graduaçao/CSTNews 6.0'
# (Verifique se o caminho é esse mesmo depois de descompactar!)

# 2. Lista para guardar os dados
dados = []

print(f"Iniciando a varredura em: {caminho_da_pasta_principal}")

# 3. "Andar" por todas as pastas e arquivos
for root, dirs, files in os.walk(caminho_da_pasta_principal):
    if root == caminho_da_pasta_principal:
        continue

    categoria = os.path.basename(root)
    print(f"Processando categoria: {categoria}...")

    # 4. Ler cada arquivo dentro da pasta da categoria
    for nome_arquivo in files:
        caminho_completo = os.path.join(root, nome_arquivo)

        try:
            # Tenta ler o arquivo (usando 'latin-1' ou 'iso-8859-1' que são comuns em datasets de notícias em português)
            with open(caminho_completo, 'r', encoding='latin-1') as f:
                texto_noticia = f.read()

            dados.append({
                'categoria': categoria,
                'nome_arquivo': nome_arquivo,
                'texto': texto_noticia
            })

        except Exception as e:
            # Se latin-1 falhar, tenta utf-8
            try:
                with open(caminho_completo, 'r', encoding='utf-8') as f:
                    texto_noticia = f.read()

                dados.append({
                    'categoria': categoria,
                    'nome_arquivo': nome_arquivo,
                    'texto': texto_noticia
                })
            except Exception as e2:
                print(f"  Erro ao ler {caminho_completo} (tentou latin-1 e utf-8): {e2}")

print("Varredura concluída!")

# 6. Criar o DataFrame do Pandas
df_dataset = pd.DataFrame(dados)

print("\n--- Amostra do DataFrame ---")
print(df_dataset.head())
print(f"\nFormato do dataset: {df_dataset.shape}")
print(f"Categorias encontradas: {df_dataset['categoria'].unique()}")

# 7. Salvar o dataset em um arquivo CSV único
nome_arquivo_final = 'cstNews_dataset_completo.csv'
df_dataset.to_csv(nome_arquivo_final, index=False)

print(f"\nSucesso! Dataset salvo como '{nome_arquivo_final}' (na pasta /content/ do Colab)")
print("Você pode baixar este arquivo no painel à esquerda.")
```

```python
from datasets import load_dataset

# Carregar o dataset CSTNews a partir do arquivo CSV gerado
# O arquivo cstNews_dataset_completo.csv foi salvo em /content/ na célula anterior.
dataset = load_dataset("csv", data_files="/content/cstNews_dataset_completo.csv")

# Exemplo de como acessar os dados (agora o dataset deve ser um DatasetDict com um 'train' split)
print(dataset['train'][0])
print(f"Dataset successfully loaded with {len(dataset['train'])} examples.")
```

```python
from transformers import BartForConditionalGeneration, BartTokenizer, Trainer, TrainingArguments, DataCollatorForSeq2Seq
from datasets import load_dataset # Ensure load_dataset is available

# 1. Initialize tokenizer and model
# You can choose a suitable BART model here. 'facebook/bart-base' is a general option.
# If you need a Portuguese-specific model, you might need to find one.
model_name = "facebook/bart-base" # Or another suitable BART model
tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)

# Tokenizar os dados
def tokenize_function(exemplos):
    # Use 'texto' column for the input (document)
    inputs = [doc for doc in exemplos["texto"]]
    model_inputs = tokenizer(inputs, max_length=1024, truncation=True)

    # The dataset (as currently structured) does not have a 'summary' column.
    # For now, as a placeholder to avoid KeyError, I'll use 'texto' again.
    # IMPORTANT: FOR PROPER SUMMARIZATION TRAINING, YOU MUST HAVE ACTUAL
    # DOCUMENT-SUMMARY PAIRS. This current mapping is a placeholder for error resolution.
    with tokenizer.as_target_tokenizer():
        labels = tokenizer(exemplos["texto"], max_length=128, truncation=True)

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# Split the main dataset into training and evaluation sets
# The 'dataset' from the previous cell is a DatasetDict with a 'train' split.
# We need to extract the 'train' split and then split it further.
dataset_splits = dataset['train'].train_test_split(test_size=0.1, seed=42)
train_dataset = dataset_splits['train']
eval_dataset = dataset_splits['test'] # Using 'test' as validation

# Tokenize the training and evaluation datasets
tokenized_train_dataset = train_dataset.map(tokenize_function, batched=True)
tokenized_eval_dataset = eval_dataset.map(tokenize_function, batched=True)

# Configurar os argumentos de treinamento
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    num_train_epochs=3,
    weight_decay=0.01,
    save_total_limit=3,
    report_to="none", # Disable Weights & Biases logging
)

# Initialize Data Collator
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

# Criar o Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_eval_dataset,
    data_collator=data_collator, # Add data_collator here
)

# Treinar o modelo
trainer.train()
```

```python
# Avaliar o modelo
results = trainer.evaluate()
print(results)
```
